```python
from pynq import Overlay
import time

# ============================================================
# 1. Overlay 路徑
# ============================================================

BITSTREAM_PATH = (
    "/home/xilinx/jupyter_notebooks/"
    "hello_pynq/hello_pynq.bit"
)

# ============================================================
# 2. 載入 Overlay
# ============================================================

print("Loading overlay...")

ol = Overlay(
    BITSTREAM_PATH,
    download=True
)

print("Overlay loaded:", ol.is_loaded())
print("Available IPs:", list(ol.ip_dict.keys()))

if not ol.is_loaded():
    raise RuntimeError("Overlay 沒有成功載入")

if "axi_gpio_0" not in ol.ip_dict:
    raise RuntimeError("Overlay 中找不到 axi_gpio_0")

# ============================================================
# 3. 取得 AXI GPIO
# ============================================================

gpio = ol.axi_gpio_0

# AXI GPIO Channel 1 registers
GPIO_DATA = 0x00
GPIO_TRI  = 0x04

# GPIO_TRI:
# bit = 0 表示 output
# bit = 1 表示 input
gpio.write(GPIO_TRI, 0x00000000)

# 先關閉全部 LED
gpio.write(GPIO_DATA, 0x00000000)

# ============================================================
# 4. 檢查暫存器
# ============================================================

data_value = gpio.read(GPIO_DATA)
tri_value = gpio.read(GPIO_TRI)

print("GPIO DATA =", hex(data_value))
print("GPIO TRI  =", hex(tri_value))

if (tri_value & 0xF) != 0:
    raise RuntimeError(
        "GPIO bit 0～3 沒有成功設定成 output"
    )

# ============================================================
# 5. 逐顆測試 LD0～LD3
# ============================================================

patterns = [
    (0b0001, "LD0"),
    (0b0010, "LD1"),
    (0b0100, "LD2"),
    (0b1000, "LD3"),
]

print("\nTesting individual LEDs...")

for value, name in patterns:
    print("Turning on", name)
    gpio.write(GPIO_DATA, value)
    time.sleep(1)

# ============================================================
# 6. 四顆 LED 閃爍
# ============================================================

print("\nBlinking all LEDs...")

for count in range(50):
    print("Blink", count + 1)

    gpio.write(GPIO_DATA, 0b1111)
    time.sleep(0.5)

    gpio.write(GPIO_DATA, 0b0000)
    time.sleep(0.5)
```
